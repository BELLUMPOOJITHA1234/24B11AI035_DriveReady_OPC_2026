```python
from langchain_core.tools import tool

catalogue = {
    "black and white printing": "₹2 per printed side",
    "colour printing": "₹10 per printed side",
    "spiral binding": "₹30 per copy"
}

@tool
def get_price(service: str) -> str:
    """Look up the unit price of black and white printing, colour printing, or spiral binding."""
    name = service.strip().lower()
    return catalogue.get(name, "This service has no listed price.")

@tool
def is_service_available(service: str) -> bool:
    """Check whether the Campus Print Shop offers a service in its catalogue."""
    name = service.strip().lower()
    return name in catalogue

# Test each tool directly
print("Colour printing:", get_price.invoke({"service": "colour printing"}))
print("Spiral binding offered:", is_service_available.invoke({"service": "spiral binding"}))
print("Lamination offered:", is_service_available.invoke({"service": "lamination"}))
```

```python
from typing import Literal
from pydantic import BaseModel, Field

class PrintOrder(BaseModel):
    """Details extracted from a customer's print request."""

    customer_name: str | None = Field(
        default=None, description="Customer's name, if given."
    )
    pages_per_copy: int | None = Field(
        default=None, ge=1, description="Number of pages in one copy."
    )
    copies: int | None = Field(
        default=None, ge=1, description="Number of copies requested."
    )
    print_type: Literal["black and white", "colour", "mixed"] | None = Field(
        default=None, description="Use mixed if the order includes both types."
    )
    spiral_binding: bool | None = Field(
        default=None, description="True if requested, False if declined, None if unspecified."
    )

# Check the record using one sample order
sample = PrintOrder(
    customer_name="Poojitha",
    pages_per_copy=20,
    copies=2,
    spiral_binding=True
)

print(sample.model_dump_json(indent=2))
```

```python
from google.colab import userdata
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=userdata.get("GROQ_API_KEY"),
    temperature=0
)

extract_order = llm.with_structured_output(PrintOrder, method="function_calling")

instructions = """
Extract a print order from the customer's message.
Use only details the customer gives. Leave missing fields as null.
'b&w' and 'bw' mean black and white. 'clr' means colour.
If some pages are colour and others are black and white, use mixed.
For spiral_binding, use true if requested, false if declined,
and null if the customer says nothing about binding.
"""

messages = [
    "poojitha here.. 20 pages b&w, 2 copies pls, spiral bind both",
    "ravi here need 1 copy of my 8 pg clr project. no binding pls",
    "hey can u print 3 copies of my 12 page assignment?",
    "sneha here, 10 pages total: colour cover + 9 bw pages, one copy",
    "need spiral binding pls, will send page count later"
]

records = []

for number, message in enumerate(messages, start=1):
    record = extract_order.invoke([
        ("system", instructions),
        ("human", message)
    ])
    records.append(record)

    print(f"\nMessage {number}: {message}")
    print(record.model_dump_json(indent=2))
```

```python
@tool
def log_print_order(order: PrintOrder) -> str:
    """Save one print order to the Orders tab in Google Sheets."""
    row = [
        order.customer_name,
        order.pages_per_copy,
        order.copies,
        order.print_type,
        order.spiral_binding
    ]
    orders_sheet.append_row(
        ["" if value is None else value for value in row],
        value_input_option="RAW"
    )
    return "Order saved to Google Sheets."

# Save the first order you extracted earlier
print(log_print_order.invoke({"order": records[0].model_dump()}))
```

```python
from langchain_core.tools import tool

@tool
def search_print_shop_facts(question: str) -> str:
    """Search shop facts about printing prices, paper usage, spiral binding,
    available services, order details, and pickup policies."""
    results = collection.query(query_texts=[question], n_results=3)
    return "\n".join(results["documents"][0])

# Test the tool with one customer question
print(search_print_shop_facts.invoke({
    "question": "How many sheets do I need for 21 double-sided pages?"
}))
```

```python
from google.colab import userdata
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver


llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=userdata.get("GROQ_API_KEY"),
    temperature=0
)

agent = create_agent(
    model=llm,
    tools=[get_price, is_service_available, search_print_shop_facts],
    checkpointer=InMemorySaver(),
    system_prompt=(
        "You are the Campus Print Shop assistant. "
        "Use the price tool for prices and the availability tool for services. "
        "Use search_print_shop_facts for shop policies and paper usage. "
        "If the retrieved facts do not answer the question, say you will "
        "check with the shop owner. Use the current conversation for customer details."
    )
)

print("Day 5 agent is ready.")
```

```python
def ask_customer(thread_id, question):
    result = agent.invoke(
        {"messages": [{"role": "user", "content": question}]},
        config={"configurable": {"thread_id": thread_id}}
    )
    return result["messages"][-1].content

print(ask_customer("customer_1", "Hi, my name is Poojitha."))
print("Customer 1:", ask_customer("customer_1", "What is my name?"))
print("Customer 2:", ask_customer("customer_2", "What is my name?"))
```

```python
import re
from uuid import uuid4

# Fresh threads keep earlier conversations from affecting this test.
run_id = uuid4().hex[:8]
results = []

def check(label, question, thread, rule):
    answer = ask_customer(f"{run_id}-{thread}", question)
    passed = rule(answer.casefold())
    results.append(passed)
    print(f"{'PASS' if passed else 'FAIL'} | {label}")
    print(answer, "\n")

def contains_number(number):
    return lambda answer: bool(re.search(rf"\b{number}\b", answer))

check("Black-and-white rate", "What is the price per black-and-white printed side?",
      "bw", contains_number(2))
check("Colour rate", "What is the price per colour printed side?",
      "colour", contains_number(10))
check("Binding rate", "What is the spiral binding price per copy?",
      "binding", contains_number(30))
check("Paper usage", "How many sheets are needed for 21 double-sided pages?",
      "paper", contains_number(11))
check("Two bindings", "What is the binding charge alone for two spiral-bound copies?",
      "two-bindings", contains_number(60))

check("Unknown Sunday hours", "What time does the shop close on Sundays?",
      "hours", lambda a: "owner" in a and not re.search(
          r"\b\d{1,2}:\d{2}\b|\b\d{1,2}\s*(?:am|pm)\b", a))

# Introduce a different customer in each thread.
ask_customer(f"{run_id}-A", "My name is Poojitha.")
ask_customer(f"{run_id}-B", "My name is Ravi.")

check("Customer A memory", "What is my name?", "A",
      lambda a: "poojitha" in a and "ravi" not in a)
check("Customer B memory", "What is my name?", "B",
      lambda a: "ravi" in a and "poojitha" not in a)

passed = sum(results)
print(f" pass rate: {passed}/{len(results)} ({passed / len(results):.1%})")
```
